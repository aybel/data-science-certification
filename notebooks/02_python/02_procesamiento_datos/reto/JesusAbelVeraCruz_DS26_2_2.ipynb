```python
## Importamos librerias
import pandas as pd
import numpy as np
from datetime import date
```

```python
# Cargamos  los datos con pandas y read_excel
df_original=pd.read_excel("hospitalData.xlsx")
```

```python

# Vemos lo que contiene el df
print("=========================Data frame Cargado==========================")
print(f"El data set contiene {df_original.shape[0]} registros")
print(f"El data set contiene {df_original.shape[1]} columnas")
print("Tipos de datos de cada columna ")
df_original.dtypes
```

```python

## Elimando los registros del codigo geografico Geographic aggregation y cambiamos el campo week ending date a date
# Generamos el df_procesado para no perder el original
df_procesado=df_original[df_original['Geographic aggregation'] != 'MP'].copy()
# En este paso cambiamos el tipo de dato de la columna de fecjas
df_procesado['Week Ending Date']=pd.to_datetime(df_procesado['Week Ending Date'], format='mixed')
print("Tipos de datos de cada columna del nuevo data frame")
df_procesado.dtypes
```

```python
## Para las columnas que tienen que ver con la ICU, cambia los valores faltantes por 0.
## las columnas de df que tiene que ver con ICU son:
#Number of ICU Beds,
#Number of Adult ICU Beds,
#Number of Pediatric ICU Beds,
#Number of ICU Beds Occupied,
#Number of Adult ICU Beds Occupied,
#Number of Pediatric ICU Beds Occupied
## son 6 columnas
df_procesado.fillna({
    'Number of ICU Beds':0,
    'Number of Adult ICU Beds':0,
    'Number of Pediatric ICU Beds':0,
    'Number of ICU Beds Occupied':0,
    'Number of Adult ICU Beds Occupied':0,
    'Number of Pediatric ICU Beds Occupied':0
},inplace=True)
print("Tipos de datos de cada columna del nuevo data set")
print("Las columnas que tienen que ver con ICU ya no tienen nulos por lo tanto la cuenta de cada columna en No-Null es la misma que la del todo el dataframe")
df_procesado.info()
```

```python
#Conoce el periodo de registro que se encuentra en el archivo para saber cuántos días han transcurrido entre la fecha más reciente y la más antigua.
fecha_antigua = df_procesado['Week Ending Date'].min()
fecha_reciente = df_procesado['Week Ending Date'].max()
# Fechas mas antigua y mas reciente
print(f"Fecha más antigua: {fecha_antigua}")
print(f"Fecha más reciente: {fecha_reciente}")
dias_transcurridos=(fecha_reciente-fecha_antigua).days
# Calculamos los días transcurridos
print(f"Dias transcurridos: {dias_transcurridos}")
```

```python
#Obtén la cantidad total de pacientes pediátricos hospitalizados con COVID-19.

# Para este total debemos sumar la siguiente columna:
# Number of Pediatric Patients Hospitalized with COVID-19: Número de pacientes pediátricos actualmente hospitalizados debido a COVID-19.
total_pacientes_pediatricos=df_procesado['Number of Pediatric Patients Hospitalized with COVID-19'].sum()
print(f"Total de pacientes pediatricos hospitalizados con COVID-19 : {total_pacientes_pediatricos}")
```

```python
#Obtén el promedio de pacientes adultos hospitalizados por RSV.
# Number of Adult Patients Hospitalized with RSV: Número de pacientes adultos actualmente hospitalizados por VRS.
promedio_pacientes_adultos_rsv=df_procesado['Number of Adult Patients Hospitalized with RSV'].mean()
print(f"Promedio de pacientes adultos hospitalizados por RSV : {promedio_pacientes_adultos_rsv}")
```

```python
#Agrupa por región geográfica y obtén la sumatoria de pacientes adultos hospitalizados con influenza.
# Para obtener esta sumatoria ahy que agrupar
# Geographic aggregation: Región geográfica para la que se agregan los datos.
# Number of Adult Patients Hospitalized with Influenza: Número de pacientes adultos actualmente hospitalizados debido a Influenza.

pacientes_por_region=df_procesado.groupby('Geographic aggregation')['Number of Adult Patients Hospitalized with Influenza'].sum().reset_index()

# Ordenamos de mayor a menor para ver las regiones más afectadas
pacientes_por_region = pacientes_por_region.sort_values(by='Number of Adult Patients Hospitalized with Influenza', ascending=False)

print("Pacientes adultos hospitalizados con influenza por region geografica")
pacientes_por_region.style.hide(axis="index")


```

```python
#Crea una columna derivada llamada "Percent of Adult ICU Beds Occupied" con el porcentaje de ocupación para las camas de los adultos en la ICU.
# Debemos usar la columna
#Number of Adult ICU Beds Occupied: Número de camas de UCI de adultos ocupadas actualmente.
df_procesado['Percent of Adult ICU Beds Occupied'] = (df_procesado['Number of Adult ICU Beds Occupied']/ df_procesado['Number of Adult ICU Beds'])*100
df_procesado.head(5)
```

```python
#Crea un *dataframe* llamado "filtro", con las filas de los hospitales que tengan más de 300 camas "disponibles" para adultos en la ICU;
#y las columnas de identificación geográfica y las relacionadas, con la cantidad de pacientes con alguna de las enfermedades respiratorias de los adultos.

#Columnas a usar:
#Para los hospitales => #Geographic aggregation: Región geográfica para la que se agregan los datos.
#Para camas disponibles para adultos en la ICU => #Number of ICU Beds: Número total de camas de UCI disponibles en la región.
# Identificacion geografica: #Geographic aggregation: Región geográfica para la que se agregan los datos.
# Cantidad de pacientes con alguna de las enfermedades respiratorias:
  # Number of Adult Patients Hospitalized with COVID-19: Número de pacientes adultos actualmente hospitalizados debido a COVID-19.
  # Number of Adult Patients Hospitalized with Influenza: Número de pacientes adultos actualmente hospitalizados debido a Influenza
  # Number of Adult Patients Hospitalized with RSV: Número de pacientes adultos actualmente hospitalizados por VRS.

#creamos el dataFrame
filtro = df_procesado[
    (df_procesado['Number of Adult ICU Beds'] > 300)
][[
    'Geographic aggregation',
    'Number of Adult ICU Beds',
    'Number of Adult Patients Hospitalized with COVID-19',
    'Number of Adult Patients Hospitalized with Influenza',
    'Number of Adult Patients Hospitalized with RSV'
]].copy()

# Mostramos las primeras filas 5 para verificar que funcionó
print("Hospitales con mas de 300 camas disponibles para adultos el al ICU")
filtro.head(5)
```

```python
#Crea un *dataframe* llamado "registros2024" que tenga todos los registros del año 2024.

#Creamis el dataframe
registros2024=df_procesado[df_procesado['Week Ending Date'].dt.year == 2024]
print("Data set del año 2024")
registros2024.head(5)
```

```python
#Crear una columna derivada con el conteo de la cantidad de certificaciones que tiene cada hospital; es decir, cuenta la cantidad de elementos que hay en la columna de certificaciones separándolas con /
df_procesado['count_certifications']=df_procesado['Certifications'].fillna('').str.count('/') + 1
print("Nueva columna count_certifications")
df_procesado.head(5)
```

```python
#Busca todos los hospitales que cumplan con la certificación CRN y déjalos en un *dataframe* llamado "CRNData".

#Creamos el dataframe
CRNData=df_procesado[df_procesado['Certifications'].str.contains('CRN', na=False)].copy()
print("Hospitales con certificacion CRN")
CRNData.head(5)
```

```python
#Almacena el *dataframe* modificado y los *dataframes* creados en los pasos anteriores en un mismo archivo de Excel llamado "datosHospitalesProcesados.xlsx". Cada *dataframe* deberá estar en una hoja diferente.
with pd.ExcelWriter('datosHospitalesProcesados.xlsx') as writer:
 df_procesado.to_excel(writer, sheet_name="modificado")
 filtro.to_excel(writer, sheet_name="filtro")
 registros2024.to_excel(writer, sheet_name="registros2024")
 CRNData.to_excel(writer, sheet_name="CRNData")
```
