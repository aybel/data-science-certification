```python
#importamos librerias necesarias pandas y pymysql
import pandas as pd
import os
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
from pandasql import sqldf
```

```python
#Generamos la url de conexioion con Mysql
import os
from dotenv import load_dotenv, find_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

load_dotenv(find_dotenv(), override=True)  # busca el .env en las carpetas padre

url = URL.create(
    "mysql+pymysql",
    username=os.environ["MYSQL_USER"],
    password=os.environ["MYSQL_PASSWORD"],
    host=os.getenv("MYSQL_HOST", "127.0.0.1"),
    port=int(os.getenv("MYSQL_PORT", "3306")),
    database=os.environ["MYSQL_DATABASE"],
)
engine = create_engine(url)
```

```python
# Leemos la tabla autos de la base de datoa
consulta="SELECT * FROM autos"
df_autos=pd.read_sql(consulta,engine)
df_autos
```

```python
pysqldf = lambda q: sqldf(q, globals())
resp=pysqldf("SELECT * from autos where marca='Mazda'")
```
