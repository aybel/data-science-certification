```python
# importamos librerias necesarias
import pandas as pd
```

```python
#importamos los 3 csvs con pandas

df1= pd.read_csv("data/archivo1.csv")
df2= pd.read_csv("data/archivo2.csv")
df3= pd.read_csv("data/archivo3.csv")

print("===Cargamos los DF===")
print(" DF1 ===")
print(df1.head())
print(" DF2 ===")
print(df2.head())
print(" DF3 ===")
print(df3.head())
```

```python
#Concatenas la información de los dos primeros DF
df_autos=pd.concat([df1,df2])
df_autos.reset_index(drop=True)
```

```python
#Relacionar este nuevo df con el del tercer df la relacion sera mediante merge, con el campo marca 
df_final=pd.merge(df_autos,df3,on="marca")
df_final
```
