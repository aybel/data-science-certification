# Notebook para practicar consultas en mongoDB
## Se tiene 3 colecciones Agencias, Inventario y Vedendeores
## La idea es hacer consultas complejas sobre esas 3 colecciones de documentos

```python
#importamos las librerias necesarias
import os
from dotenv import load_dotenv, find_dotenv
from pymongo import MongoClient
```

```python
#Conexión 
load_dotenv(find_dotenv(), override=True)
client = MongoClient(
    host=os.getenv("MONGO_HOST", "127.0.0.1"),
    port=int(os.getenv("MONGO_PORT", "27017")),
    username=os.environ["MONGO_USER"],
    password=os.environ["MONGO_PASSWORD"],
    authSource="admin",
)
client.admin.command("ping")
```

```python
db = client['admin']
col_agencias= db['agencias']
col_vendedores= db['vendedores']
col_inventarios= db['inventario']
col_agencias
```

```python
query={"activa":True}
agancias_activas=list(col_agencias.find(query))
agancias_activas           
```

```python
query={"activa":True}
field_projection={"nombre":1,"marcas_autorizadas":1}
agancias_activas=list(col_agencias.find(query,field_projection))
agancias_activas          
```

```python
cuantos=col_agencias.count_documents({})
cuantos
```
